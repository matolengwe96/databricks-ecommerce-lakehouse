```python
from pyspark.sql.functions import col

df_bronze_cust = spark.read.table("bronze_customers")
df_silver_cust = df_bronze_cust.dropDuplicates().withColumn("customer_id", col("customer_id").cast("string"))
df_silver_cust.write.format("delta").mode("overwrite").saveAsTable("silver_customers")

df_bronze_orders = spark.read.table("bronze_orders")
df_silver_orders = df_bronze_orders.dropDuplicates().withColumn("order_date", col("order_date").cast("date")).withColumn("total_amount_usd", col("total_amount_usd").cast("double"))
df_silver_orders.write.format("delta").mode("overwrite").saveAsTable("silver_orders")

df_bronze_prod = spark.read.table("bronze_product_summary")
df_silver_prod = df_bronze_prod.dropDuplicates().withColumn("avg_price", col("avg_price").cast("double"))
df_silver_prod.write.format("delta").mode("overwrite").saveAsTable("silver_product_summary")
```
